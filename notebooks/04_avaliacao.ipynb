## 🎯 Objetivos deste Notebook
1. **Comparar Métricas**: Analisar RMSE, MAE, Precision@10 e Recall@10 dos modelos treinados.
2. **Visualização Gráfica**: Gráficos Plotly comparativos.
3. **Justificativa Técnica**: Documentar a escolha do modelo campeão (SVD).

```python
import sys
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.templates.default = "plotly_white"

ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.config import MODELS_DIR
df_comp = pd.read_csv(MODELS_DIR / "model_comparison.csv")
display(df_comp)
```

## 1. Gráficos Comparativos de Desempenho (Plotly)

```python
fig_rmse = px.bar(
    df_comp,
    x='rmse',
    y='model_name',
    orientation='h',
    text='rmse',
    title="<b>Comparativo de Erro (RMSE) — Menor é Melhor</b>",
    color='rmse',
    color_continuous_scale='Reds_r'
)
fig_rmse.update_layout(yaxis={'categoryorder': 'total descending'}, height=500)
fig_rmse.show()

fig_prec = px.bar(
    df_comp,
    x='precision_at_10',
    y='model_name',
    orientation='h',
    text='precision_at_10',
    title="<b>Comparativo de Recomendação (Precision@10) — Maior é Melhor</b>",
    color='precision_at_10',
    color_continuous_scale='Viridis'
)
fig_prec.update_layout(yaxis={'categoryorder': 'total ascending'}, height=500)
fig_prec.show()
```

## 2. Justificativa Técnica da Escolha do Modelo Campeão

> **🏆 Modelo Campeão Selecionado: SVD (n_factors=100, n_epochs=30, lr_all=0.01, reg_all=0.1)**
>
> **Justificativa**:
> 1. **Menor Erro de Predição**: Registrou o menor **RMSE (0.9078)** e menor **MAE (0.7167)** entre todos os algoritmos avaliados.
> 2. **Maior Precisão no Top-10**: Registrou a maior **Precision@10 (0.6047)** e maior **Recall@10 (0.7375)**, garantindo que 60.4% das recomendações no Top-10 sejam itens altamente relevantes (nota $\ge 4.0$).
> 3. **Eficiência Computacional & Explicabilidade**: A representação vetorial em fatores latentes do SVD pode ser combinada com o perfil de gêneros dos filmes para gerar justificativas explicativas robustas no módulo `src/explainer.py`.
